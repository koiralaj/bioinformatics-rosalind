# **Problem 9 – Computing GC Content**

```python
# FASTA from Rosalind download; print ID and GC% of the highest-GC record
from Bio import SeqIO
from Bio.SeqUtils import gc_fraction

best_id = ""
best_gc = -1
for rec in SeqIO.parse("/content/rosalind_gc.txt", "fasta"):
    gc = gc_fraction(rec.seq) * 100
    if gc > best_gc:
        best_gc = gc
        best_id = rec.id
print(best_id)
print(best_gc)
```
